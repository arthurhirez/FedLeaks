```python
from itertools import permutations
import matplotlib.pyplot as plt

def generate_combinations(n: int):
    """
    Generate all unique concatenated permutations of n components
    consisting of:
      - 1 'H'
      - 1 'M'
      - (n-2) 'L'
    Returns list of strings.
    """
    if n < 3:
        raise ValueError("n must be at least 3 (1 H, 1 M, and >=1 L).")
    
    # Build base list
    components = ['H', 'M'] + ['L'] * (n - 2)
    
    # Generate unique permutations
    unique_combos = set(permutations(components))
    
    # Concatenate tuples into strings and sort for consistency
    combos_list = sorted([''.join(p) for p in unique_combos])
    
    return combos_list


# Example usage
n = 5
combinations = generate_combinations(n)
combinations = [c for c in combinations if c[0] != 'L']
# Print list
print(combinations)
print(f"Total unique combinations: {len(combinations)}")
```

```python
processed_detect = df_detect.copy()
processed_detect[['window', 'step_size', 'interval']] = processed_detect['setup'].str.split('_', expand = True).astype(int)

processed_detect['interval'] = (processed_detect['interval'] / 3600).astype(int)
processed_detect['step_ratio'] = round(processed_detect['step_size']  / processed_detect['window'], 3)


processed_detect['init_state'] = processed_detect['state_map'].str[:2]
processed_detect['final_state'] = processed_detect['state_map'].str[3:5]

processed_detect.drop(columns = ['setup', 'exp_id'], inplace = True)



qwe = pd.read_csv(f'results/income_base/traceback/INCOMESIMPLE/compiled_similarity.csv')
qwe['state_map'] = qwe['target_init_final_state'] + '__' + qwe['initial_state_mapping']

qwe['init_state'] = qwe['target_init_final_state'].str[:2]
qwe['final_state'] = qwe['target_init_final_state'].str[-2:]


qwe[['A', 'B', 'C', 'D', 'E']] = qwe['initial_state_mapping'].str.split('_', expand = True)

def find_similar(row):
        match_cols = [col for col in ['A', 'B', 'C', 'D', 'E'] if row[col] == row['final_state']]
        return '_'.join(match_cols)

qwe['similar_districts'] = qwe.apply(find_similar, axis=1)
qwe['ground_truth'] = qwe['target_district_node'].str[0]
qwe = qwe[['state_map', 'A', 'B', 'C', 'D', 'E', 'similar_districts']].drop_duplicates()

processed_detect = pd.merge(processed_detect, qwe,
                           on = 'state_map', how = 'left')



processed_detect['detection'] = processed_detect['correct_%']

processed_detect['period'] = processed_detect['period'].replace('all_periods', 'Anual').replace('all_year', 'Anual').replace('first_4_months', 'Inicial_4').replace('first_year', 'Ano_1').replace('second_year', 'Ano_2')

processed_detect['method'] = processed_detect['method'].replace('max cosine', 'Máximo').replace('rank cosine_mean', 'Rank_média').replace('rank cosine_sum', 'Rank_soma').replace('summary', 'Sumário')
processed_detect['Método'] = processed_detect['period'] + '__' + processed_detect['method']


qwe_aux = processed_detect.copy()
qwe_aux['similar_districts'] = 'Agrup.'

processed_detect = pd.concat([processed_detect, qwe_aux])


processed_detect.rename(columns  = {
    'step_ratio' : 'Passo',
    'epochs_tgt' : 'Epc.',
    'interval' : 'Fator Agreg.',
    'period_tgt' : 'Período',
    'final_state' : 'Estado Final',
    'window' : 'Janela',
    'similar_districts' : 'Cliente similar',
    'detection' : 'Detecção',
    }, inplace = True)

processed_detect.head()
```

```python
def plot_grid_dot_error(data, x_var, y_var, hue_var, col_var, row_var, palette,
                        col_order=None, row_order=None, ylims=None, error_type='sd'):
    """FacetGrid: Dot plot + error bars (Seaborn 0.14+ syntax, spaced dots)"""
    
    g = sns.FacetGrid(
        data=data,
        col=col_var,
        row=row_var,
        col_order=data[col_var].sort_values().unique() if col_order is None else col_order,
        row_order=data[row_var].sort_values().unique() if row_order is None else row_order,
        margin_titles=True,
        despine=False,
        height=3,
        aspect=1
    )

    g.map_dataframe(
        sns.pointplot,
        x=x_var,
        y=y_var,
        order=data[x_var].sort_values().unique(),
        hue=hue_var,
        hue_order=data[hue_var].sort_values().unique(),
        palette=palette,
        dodge=0.9,                # more spacing between hue groups
        linestyle='none',         # removes lines between points
        errorbar='sd' if error_type == 'sd' else 95,
        err_kws={'linewidth': 4}, # error bar line width
        capsize=.2,
        markers='o',
        markersize=10              
    )

    g.fig.subplots_adjust(hspace=0.25, wspace=0.1)
    g.add_legend(title=None, frameon=False, ncol=len(data[hue_var].unique().tolist()))
    g._legend.set_bbox_to_anchor((0.775, 0.51))  # adjust as needed
    
    ylims = (0, 100) if ylims is None else ylims
    for ax in g.axes.flat:
        ax.set_ylim(ylims)
        ax.tick_params(axis='x', labelsize=14)
        y_min, y_max = ax.get_ylim()
        ax.set_yticks(np.arange(y_min, y_max + 1, 25))
        ax.grid(zorder=0)
    
        # Remove x-axis ticks and label
        ax.set_xticks([])
        ax.set_xlabel('')

    plt.tight_layout()
    plt.show()
```

```python
def plot_grid(data, x_var, y_var, hue_var, col_var, row_var, palette,
             col_order = None, row_order = None, ylims = None, error_type='sd'):
    # Define FacetGrid with tighter spacing
    g = sns.FacetGrid(
        data = data,
        col=col_var,
        row=row_var,
        col_order=data[col_var].sort_values().unique() if col_order is None else col_order,
        row_order=data[row_var].sort_values().unique() if row_order is None else row_order,
        margin_titles=True,  # Helps use space better
        despine=False,  # Keeps axes for reference
        height=1.85,  # Smaller subplot height
        aspect=1  # Adjust aspect ratio (width / height)
    )

    hue_order=[
                'Anual__Máximo',
                 'Anual__Rank_média',
                 'Anual__Rank_soma',
        'Anual__Sumário',
                 'Inicial_4__Máximo',
                 'Inicial_4__Rank_média',
                 'Inicial_4__Rank_soma',
               
                'Ano_1__Máximo',
                 'Ano_1__Rank_média',
                 'Ano_1__Rank_soma',
                 'Ano_2__Máximo',
                 'Ano_2__Rank_média',
                 'Ano_2__Rank_soma',
                    ]
    
    # Map barplot with narrower bars
    g.map_dataframe(
        sns.barplot,
        x=x_var,
        y=y_var,
        order=data[x_var].sort_values().unique(),
        hue=hue_var,
        hue_order=hue_order,
        width=.8,
        palette=palette,
        zorder=25,
        errorbar='sd' if error_type == 'sd' else 95,
        # capsize=0.255,
        err_kws={'color': 'black', 'linewidth' : 1.255}
    )
    
    
    # Reduce spacing between subplots
    g.fig.subplots_adjust(
        hspace=0.25,  # vertical space between rows
        wspace=0.05  # horizontal space between columns
    )
    
    # Desired legend order
    legend_order = [
        
         'Ano_1__Máximo',
         'Ano_1__Rank_média',
         'Ano_1__Rank_soma',
         'Ano_2__Máximo',
         'Ano_2__Rank_média',
         'Ano_2__Rank_soma',
         'Anual__Máximo',
         'Anual__Rank_média',
         'Anual__Rank_soma',
         'Inicial_4__Máximo',
         'Inicial_4__Rank_média',
         'Inicial_4__Rank_soma',
        'Anual__Sumário',
        ]
    
    # After mapping the barplot, remove automatic legend
    if g._legend is not None:
        g._legend.remove()
    
    # Rebuild handles and labels in desired order
    handles, labels = [], []
    for ax in g.axes.flatten():
        h, l = ax.get_legend_handles_labels()
        handles.extend(h)
        labels.extend(l)
    
    # Deduplicate handles/labels
    seen = {}
    for handle, label in zip(handles, labels):
        if label not in seen:
            seen[label] = handle
    
    # Reorder according to your legend_order
    handles_ordered = [seen[label] for label in legend_order if label in seen]
    labels_ordered = [label for label in legend_order if label in seen]
    
    # Add legend in desired position
    g.fig.legend(
        handles_ordered, labels_ordered,
        title=hue_var,
        ncol=5,
        loc='center',
        bbox_to_anchor=(0.51, -0.10),
        frameon=True
    )

    def apply_alpha_to_bars(ax, alpha_map):
        for patch, label in zip(ax.patches, ax.get_legend_handles_labels()[1] * (len(ax.patches)//len(ax.get_legend_handles_labels()[1]))):
            target_alpha = alpha_map.get(label, 1.0)
            patch.set_alpha(target_alpha)
            
    alpha_map = {
        'Anual__Máximo': 0.6,
        'Anual__Rank_média': 0.6,
        'Anual__Rank_soma': 0.6,
        'Anual__Sumário': 1.0,
        'Inicial_4__Máximo': 1.0,
        'Inicial_4__Rank_média': 1.0,
        'Inicial_4__Rank_soma': 1.0,
        'Ano_1__Máximo': 0.25,
        'Ano_1__Rank_média': 0.25,
        'Ano_1__Rank_soma': 0.25,
        'Ano_2__Máximo': 0.25,
        'Ano_2__Rank_média': 0.25,
        'Ano_2__Rank_soma': 0.25,

    }
                
    for ax in g.axes.flat:
        if not ax.patches:
            continue
        
        # === Bars ===
        n_per_xtick = len(hue_order)
        reps = len(ax.patches) // n_per_xtick
        labels = hue_order * reps
        for patch, label in zip(ax.patches, labels):
            patch.set_alpha(alpha_map.get(label, 1.0))
    
        # === Error bars ===
        # barplot draws one LineCollection per bar group
        if ax.collections:
            reps = len(ax.collections) // n_per_xtick
            labels = hue_order * reps
            for coll, label in zip(ax.collections, labels):
                coll.set_alpha(alpha_map.get(label, 1.0))
    
        # Optionally, also adjust caps (small horizontal lines)
        for line, label in zip(ax.lines, labels):
            line.set_alpha(alpha_map.get(label, 1.0))



    ylims = (0, 100) if ylims is None else ylims
    # Optional: Rotate x-ticks and tighten layout
    for ax in g.axes.flat:
        for label in ax.get_xticklabels():
            label.set_rotation(45)
            label.set_fontsize(8)
        # ax.set_ylabel('')
        # ax.set_xlabel('')
        ax.set_ylim(ylims)
        y_min, y_max = ax.get_ylim()
        ax.set_yticks(np.arange(y_min, y_max + 1, 20))
        ax.grid(zorder = 0)
        ax.set_xticks([])
        ax.set_xlabel('')
        ax.set_ylabel('Detecção (%)')
        
        for line in ax.lines:
            line.set_zorder(30)
            
    plt.tight_layout()
    plt.savefig('results/imgs/report/Detect_hyper_income.pdf', dpi = 400, bbox_inches='tight')
    plt.show()
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Cleaner style
sns.set_theme(style="whitegrid", font_scale=1.2)

qwe_plot = plot_detect_hyper[plot_detect_hyper['Epochs'] == 5].copy()

custom_palette = {
    0.05: "#1b9e77",
    0.1:  "#d95f02",
    0.2:  "#7570b3",
    0.4:  "#e7298a"
}

# Step 1: seaborn plot with CI shading
g = sns.relplot(
    data=qwe_plot,
    x='LSTM',
    y='Detecção',
    hue='InfoNCE',
    style='Estado Final',
    style_order=["ML", "HL"],
    markers=True,
    markersize=9,
    kind='line',
    errorbar="ci",  # CI shading
    palette=custom_palette,
    linewidth=2.5,
    col='Cliente similar',
    col_order=['Geral', 'A', 'B', 'C', 'D', 'E'],
    col_wrap=3,
    height=3,
    aspect=1,
    facet_kws={'sharey': True, 'sharex': True}
)

order = ['Geral', 'A', 'B', 'C', 'D', 'E']

qwe_plot['Cliente similar'] = pd.Categorical(
    qwe_plot['Cliente similar'],
    categories=order,
    ordered=True
)

stats = (
    qwe_plot
    .groupby(["Cliente similar","LSTM","InfoNCE","Estado Final"],observed=True)["Detecção"]
    .agg(["mean","std"])
    .reset_index()
)


aux = {}
# Step 3: overlay mean - max(std) as black line
for ax, (facet, subfacet) in zip(g.axes.flatten(), stats.groupby("Cliente similar")):
    for state, sub_state in subfacet.groupby("Estado Final"):
        # linestyle consistent with style (LM vs LH)
        ls = "-" if state == "ML" else "--"

        # compute mean & max std per x
        agg = (
            sub_state.groupby("LSTM")
            .agg(mean=("mean","mean"), maxstd=("std","max"))
            .reset_index()
        )
        
        # draw only mean - maxstd
        ax.plot(
            agg["LSTM"],
            agg["mean"] - agg["maxstd"],
            color="black", linestyle=ls, linewidth=1.5,
            label=f"{state} – Max STD"   # label so it appears in legend
        )
        aux[facet + '__' + state] = agg

# Adjust axes for each facet
for ax in g.axes.flatten():
    ax.set_xticks(sorted(qwe_plot['LSTM'].unique()))
    ax.set_yticks(np.arange(0, 101, 10))
    ax.set_ylim(0, 100)
    ax.grid(axis="y", alpha=0.6)
    ax.set_xlabel("LSTM")
    ax.set_ylabel("Detecção (%)")

# Collect handles/labels from all axes
handles, labels = [], []
for ax in g.axes.flatten():
    h, l = ax.get_legend_handles_labels()
    handles.extend(h)
    labels.extend(l)

# Deduplicate
seen = {}
for h, l in zip(handles, labels):
    if l not in seen:
        seen[l] = h
handles_clean = list(seen.values())
labels_clean = list(seen.keys())

# --- Reorder here ---
# Example: put "– Max STD" at the end
maxstd_idx = [i for i, l in enumerate(labels_clean) if "Max STD" in l]
other_idx = [i for i, l in enumerate(labels_clean) if "Max STD" not in l]
new_order = other_idx + maxstd_idx

handles_ordered = [handles_clean[i] for i in new_order]
labels_ordered = [labels_clean[i] for i in new_order]

handles_ordered[-2:] = handles_ordered[-2:][::-1]
labels_ordered[-2:] = labels_ordered[-2:][::-1]


labels_ordered = list(labels_ordered)
handles_ordered = list(handles_ordered)

# Place legend below the plots
if g._legend is not None:
    g._legend.remove()

g.fig.legend(
    handles_ordered, labels_ordered,
    loc="lower center", bbox_to_anchor=(1.075, 0.35), frameon=False#, ncol = 4
)

sns.despine()
plt.tight_layout()
plt.savefig('results/imgs/report/Detect_hyper_income_districts.pdf', dpi = 400, bbox_inches='tight')
plt.show()
```

```python
import glob
import os
import pandas as pd

# Step 1: Get all matching directories
base_path = 'results/income_base/traceback/HYPER_INCOME'
# base_path = 'results/density_base/traceback/HYPER'
pattern = os.path.join(base_path, '*2_*/')
directories = [d for d in glob.glob(pattern) if os.path.isdir(d)]

# Step 2: Count files in each directory

aux_det = []
aux_sim = []

for dir_path in directories:
    files = [f for f in glob.glob(os.path.join(dir_path, '*')) if os.path.isfile(f)]
    det = pd.read_csv(f'{dir_path}/compiled_detection.csv')
    sim = pd.read_csv(f'{dir_path}/compiled_similarity.csv')

    aux_det.append(det)
    aux_sim.append(sim)

df_detect_hyper = pd.concat(aux_det, ignore_index=True)
df_detect_hyper.head()
```

```python
processed_detect_hyper = df_detect_hyper.copy()
processed_detect_hyper[['window', 'step_size', 'interval', 'lstm', 'info']] = processed_detect_hyper['setup'].str.split('_', expand = True).astype(float)

for c in ['window', 'step_size', 'interval', 'lstm']:
    processed_detect_hyper[c] = processed_detect_hyper[c].astype(int)
    
processed_detect_hyper['interval'] = (processed_detect_hyper['interval'] / 3600).astype(int)
processed_detect_hyper['step_ratio'] = round(processed_detect_hyper['step_size']  / processed_detect_hyper['window'], 3)


processed_detect_hyper['init_state'] = processed_detect_hyper['state_map'].str[:2]
processed_detect_hyper['final_state'] = processed_detect_hyper['state_map'].str[3:5]

processed_detect_hyper.drop(columns = ['setup', 'exp_id', 'state_map'], inplace = True)

for c in ['window', 'step_size', 'interval', 'lstm', 'info']:
    print(c, processed_detect_hyper[c].unique().tolist())
    
processed_detect_hyper.head()
```

```python
plot_detect_hyper = processed_detect_hyper.groupby(['final_state', 'window', 'lstm', 'info'] 
                                                              ).agg({
                                                                'correct_%' : 'sum',
                                                                'period' : 'count'
                                                            }).rename(columns = {'period' : 'count'}).reset_index()

plot_detect_hyper['detection'] = plot_detect_hyper['correct_%'] / plot_detect_hyper['count']

x_var = 'lstm'
y_var = 'detection'
hue_var = 'info'

col_var = 'window'
row_var = 'final_state'

row_order = ['ML', 'HL']

plot_grid(data = plot_detect_hyper, x_var = x_var, y_var = y_var, hue_var = hue_var, col_var = col_var, row_var = row_var,
             col_order = None, row_order = row_order, ylims = (30,105))
```

```python
processed_detect_hyper_selected = processed_detect_hyper[(processed_detect_hyper['lstm'] == 30) & (processed_detect_hyper['info'] == 0.1)]

processed_detect_hyper_selected = processed_detect_hyper_selected[(processed_detect_hyper_selected['period_tgt'] == 'Targeted') & 
                                                            (processed_detect_hyper_selected['epochs_tgt'] == 'all') & 
                                                            (processed_detect_hyper_selected['skip_tgt'] == True)]


plot_detect_hyper_selected = processed_detect_hyper_selected.groupby(['final_state', 'epochs_tgt', 'skip_tgt',  # epocas utilizadas
                                                                 'period', 'period_tgt',    # meses (labels) considerados
                                                                 'drift_type', 'method']  # tipo análise
                                                              ).agg({
                                                                'correct_%' : 'sum',
                                                                'period' : 'count'
                                                            }).rename(columns = {'period' : 'count'}).reset_index()

plot_detect_hyper_selected['detection'] = plot_detect_hyper_selected['correct_%'] / plot_detect_hyper_selected['count']


x_var = 'drift_type'
y_var = 'detection'
hue_var = 'period'

col_var = 'method'
row_var = 'final_state'
row_order = ['ML', 'HL']

plot_grid(data = plot_detect_hyper_selected, x_var = x_var, y_var = y_var, hue_var = hue_var, col_var = col_var, row_var = row_var,
             col_order = None, row_order = row_order, ylims = (50,105))
```
